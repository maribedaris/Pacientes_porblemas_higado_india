## 📚 Importar librerías


```python
from pathlib import Path

import pandas as pd
```

## 💾 Descargar y guardar datos RAW


```python
URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/ilpd/Indian%20Liver%20Patient%20Dataset%20%28ILPD%29.csv"
COLUMNAS = [
    "age",
    "gender",
    "total_bilirubin",
    "direct_bilirubin",
    "alkaline_phosphotase",
    "alamine_aminotransferase",
    "aspartate_aminotransferase",
    "total_proteins",
    "albumin",
    "albumin_and_globulin_ratio",
    "dataset",
]

def find_project_root(start: Path) -> Path:
    for path in [start, *start.parents]:
        if (path / ".git").exists():
            return path
    raise FileNotFoundError("No se encontró la raíz del proyecto (.git)")

project_root = find_project_root(Path.cwd())
raw_path = project_root / "data" / "01_raw" / "indian_liver_patient_dataset.csv"

df_raw = pd.read_csv(URL, header=None, names=COLUMNAS)
raw_path.parent.mkdir(parents=True, exist_ok=True)
df_raw.to_csv(raw_path, index=False)

print(f"Archivo RAW guardado en: {raw_path}")
print(f"Filas: {df_raw.shape[0]}, Columnas: {df_raw.shape[1]}")
```

## 🧠 Entendimiento del problema

### ¿Cuál es el objetivo del problema?
Clasificar si un paciente presenta enfermedad hepática o no a partir de variables clínicas y de laboratorio.

### ¿Cómo se usará su solución?
Como apoyo a la toma de decisiones médicas (priorización y tamizaje), no como diagnóstico definitivo.

### ¿Cuáles son las soluciones actuales (si las hay)?
Evaluación clínica por especialistas, interpretación manual de exámenes de laboratorio y reglas médicas.

### ¿Cómo se debe enmarcar este problema?
Problema de aprendizaje supervisado de clasificación binaria, ejecutado fuera de línea para entrenamiento y luego en línea/batch para inferencia según el despliegue.

### ¿Cómo se debe medir el desempeño o rendimiento, una primera intuición?
Con métricas de clasificación: recall (sensibilidad), precision, F1 y ROC-AUC; dando prioridad a recall para no dejar pacientes enfermos sin detectar.

### ¿La medida de desempeño está alineada con el objetivo del problema?
Sí. Priorizar recall se alinea con reducir falsos negativos en un contexto de salud.

### ¿Cuál sería el desempeño mínimo necesario para alcanzar el objetivo?
Como línea base inicial: recall >= 0.80 y ROC-AUC superior al baseline simple (por ejemplo, clasificación mayoritaria).

### ¿Cuáles son los problemas parecidos? ¿Se puede reutilizar experiencias o herramientas?
Problemas de clasificación clínica (diabetes, riesgo cardiovascular, enfermedad renal). Se pueden reutilizar pipelines de clasificación tabular, validación cruzada y calibración de probabilidades.

### ¿Hay experiencia del problema disponible?
Sí, existen publicaciones y ejercicios académicos previos sobre ILPD en clasificación supervisada.

### (Importante) ¿Cómo se puede resolver el problema manualmente?
Un médico revisa síntomas, antecedentes y resultados de laboratorio (bilirrubina, enzimas hepáticas, proteínas, etc.) para determinar probabilidad de enfermedad hepática y solicitar estudios adicionales.
